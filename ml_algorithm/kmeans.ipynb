```python
import numpy as np

class KMeans:
    def __init__(self, k=3, max_iters=100):
        self.k = k
        self.max_iters = max_iters
        self.centroids = None

    def fit(self, X):
        # 1. Initialize: Pick K random points from X to be the starting centroids
        random_indices = np.random.choice(X.shape[0], self.k, replace=False)
        self.centroids = X[random_indices]

        for _ in range(self.max_iters):
            # 2. Assign: Find the closest centroid for every data point
            labels = self._assign_clusters(X)
            
            # Save the old centroids to check if they stop moving later
            old_centroids = self.centroids.copy()
            
            # 3. Update: Calculate the new mean (center of mass) for each cluster
            for i in range(self.k):
                # Get all points assigned to cluster 'i'
                cluster_points = X[labels == i]
                if len(cluster_points) > 0:
                    self.centroids[i] = np.mean(cluster_points, axis=0)
            
            # 4. Repeat until Convergence: If the centroids didn't move, we are done!
            if np.all(old_centroids == self.centroids):
                break

    def _assign_clusters(self, X):
        # Calculate distance from each point to each centroid, pick the minimum
        labels = []
        for x in X:
            distances = [np.linalg.norm(x - centroid) for centroid in self.centroids]
            labels.append(np.argmin(distances)) # Returns the index of the closest centroid
        return np.array(labels)
```
