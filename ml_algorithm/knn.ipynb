```python
import numpy as np
from collections import Counter

class KNN:
    def __init__(self, k=3):
        self.k = k

    def fit(self, X, y):
        # KNN doesn't "train". It just stores the training data.
        self.X_train = X
        self.y_train = y

    def predict(self, X):
        # We need to predict the label for every single row in the test set X
        predicted_labels = [self._predict_single(x) for x in X]
        return np.array(predicted_labels)

    def _predict_single(self, x):
        # 1. Calculate Euclidean distances between this point 'x' and ALL training points
        # np.linalg.norm computes the Euclidean distance automatically
        distances = [np.linalg.norm(x - x_train) for x_train in self.X_train]
        
        # 2. Sort by distance and get the indices of the first 'k' neighbors
        # np.argsort returns the indices that would sort the array
        k_indices = np.argsort(distances)[:self.k]
        
        # 3. Extract the labels of those 'k' nearest neighbor training samples
        k_nearest_labels = [self.y_train[i] for i in k_indices]
        
        # 4. Take a majority vote (find the most common label)
        most_common = Counter(k_nearest_labels).most_common(1)
        return most_common[0][0]
```
