```python
import numpy as np
import matplotlib.pyplot as plt

class LinearRegression:
    def __init__(self, learning_rate=0.01, n_iterations=1000):
        self.lr = learning_rate
        self.n_iterations = n_iterations
        self.weights = None
        self.bias = None
        self.loss_history = [] # Bonus points: track loss to show it's learning

    def fit(self, X, y):
        # 1. Extract number of samples (n_samples) and features (n_features)
        n_samples, n_features = X.shape
        
        # 2. Initialize parameters (weights to zeros, bias to 0)
        self.weights = np.zeros(n_features)
        self.bias = 0
        
        # 3. Gradient Descent Loop
        for _ in range(self.n_iterations):
            
            # Forward pass: Calculate predictions
            # y_bar = X * w + b can be efficiently computed using np.dot for matrix multiplication
            
            y_predicted = np.dot(X, self.weights) + self.bias
            
            # Calculate the error vector
            error = y_predicted - y
            
            # Record the Mean Squared Error (for debugging/monitoring)
            mse = (1 / (2 * n_samples)) * np.sum(error ** 2)
            self.loss_history.append(mse)
            
            # Backward pass: Calculate gradients using vectorized operations
            # np.dot handles the matrix multiplication: X.T dot error
            dw = (1 / n_samples) * np.dot(X.T, error)
            db = (1 / n_samples) * np.sum(error)
            
            # Update the parameters
            self.weights -= self.lr * dw
            self.bias -= self.lr * db

    def predict(self, X):
        # Make final predictions using the trained weights
        return np.dot(X, self.weights) + self.bias
```
